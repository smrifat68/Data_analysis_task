```python
def missing_pct(df):
    # Calculate missing value and their percentage for each column
    missing_count_percent = df.isnull().sum() * 100 / df.shape[0]
    df_missing_count_percent = pd.DataFrame(missing_count_percent).round(2)
    df_missing_count_percent = df_missing_count_percent.reset_index().rename(
                    columns={
                            'index':'Column',
                            0:'Missing_Percentage (%)'
                    }
                )
    df_missing_value = df.isnull().sum()
    df_missing_value = df_missing_value.reset_index().rename(
                    columns={
                            'index':'Column',
                            0:'Missing_value_count'
                    }
                )
    # Sort the data frame
    #df_missing = df_missing.sort_values('Missing_Percentage (%)', ascending=False)
    Final = df_missing_value.merge(df_missing_count_percent, how = 'inner', left_on = 'Column', right_on = 'Column')
    Final = Final.sort_values(by = 'Missing_Percentage (%)',ascending = False)
    return Final

missing_pct(df)
```

```python
df['rating'] = df['rating'].replace({'74 min': 'TV-MA', '84 min': 'TV-MA', '66 min': 'TV-MA'})
df['rating'] = df['rating'].replace({'TV-Y7-FV': 'TV-Y7'})
print(df.rating.unique())
```

```python
# Renaming vaules for Rating for better understanding
# Source : https://help.netflix.com/en/node/2064
df['rating'] = df['rating'].replace({
                'PG-13': 'Teens - Age above 12',
                'TV-MA': 'Adults',
                'PG': 'Kids - with parental guidence',
                'TV-14': 'Teens - Age above 14',
                'TV-PG': 'Kids - with parental guidence',
                'TV-Y': 'Kids',
                'TV-Y7': 'Kids - Age above 7',
                'R': 'Adults',
                'TV-G': 'Kids',
                 'G': 'Kids',
                'NC-17': 'Adults',
                'NR': 'NR',
                'UR' : 'UR'
                
})
```

```python
df['rating'].unique()
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])

df['cast'].replace(np.nan, 'No Data',inplace  = True)
df['director'].replace(np.nan, 'No Data',inplace  = True)
df.dropna(inplace=True)

# Drop Duplicates
df.drop_duplicates(inplace= True)

print(df.shape)
```

```python
# Breakdown of Content Type (Movies vs TV Shows)
print("\nContent Type Distribution")
print(df['type'].value_counts())
```

```python
# 1. Preprocessing Duration for Movies (minutes) and TV Shows (seasons)
df_movies = df[df['type'] == 'Movie'].copy()
df_movies['duration_min'] = df_movies['duration'].str.replace(' min', '').astype(float)

# 2. Check correlation between Release Year and Movie Duration
correlation_matrix = df_movies[['release_year', 'duration_min']].corr()
print("--- Correlation Matrix (Movies) ---")
print(correlation_matrix)

```

```python
# 3. Visualize a heatmap (if more numerical features are engineered)
plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation: Release Year vs Movie Duration')
plt.show()
```

```python
# Convert 'date_added' to datetime and extract the year Netflix added the content
df['year_added'] = pd.to_datetime(df['date_added'].str.strip()).dt.year

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='year_added', hue='type', palette='Set1')
plt.title('Content Added to Netflix Over the Years')
plt.xlabel('Year Added')
plt.ylabel('Count of Titles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
# Splitting in case of multiple country listings, taking the primary country
top_countries = df['country'].str.split(', ').str[0].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
# Exploding genres since titles can belong to multiple genres
genres = df['listed_in'].str.split(', ').explode().value_counts().head(10)
sns.barplot(x=genres.values, y=genres.index, palette='mako')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Count')
plt.show()
```

```python
# Total Shows and movies

df_count = df['show_id'].count().sum()
print(df_count)
# Split of showes and TV
df_type = df.groupby('type')['show_id'].count().reset_index()
df_type = df_type.rename(columns = {"show_id":"count_showids"})
df_type
```

```python

# Set the style grid
sns.set_theme(style="whitegrid")

# 1. Create a figure canvas with 1 row and 2 columns of plots
# figsize=(15, 6) makes the window wide enough to fit both nicely side-by-side
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# ==========================================
# LEFT PLOT: Pie Chart (goes to axes[0])
# ==========================================
colors = ['#E50914', '#564d4d']
explode = (0.05, 0)

# Notice we use 'axes[0].pie' instead of 'plt.pie'
patches, texts, autotexts = axes[0].pie(
    df_type['count_showids'], 
    labels=df_type['type'], 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colors, 
    explode=explode,
    textprops={'fontsize': 12}
)

# Style percentages inside the pie slice
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_weight('bold')

# Set title specifically for the first plot
axes[0].set_title('Proportion of Movies vs TV Shows', fontsize=14, fontweight='bold', pad=15)


# ==========================================
# RIGHT PLOT: Bar Plot (goes to axes[1])
# ==========================================
# Crucial: We pass 'ax=axes[1]' to tell Seaborn to draw inside the second subplot space
sns.barplot(x='type', y='count_showids', data=df_type, palette=['#E50914', '#221F1F'], ax=axes[1])

# Set labels and title for the second plot using the axes syntax (set_title, set_xlabel)
axes[1].set_title('Number of Titles by Type', fontsize=14, fontweight='bold', pad=15)
axes[1].set_xlabel('Content Type', fontsize=12)
axes[1].set_ylabel('Count', fontsize=12)

# Dynamic data labels on top of the bars
for p in axes[1].patches:
    axes[1].annotate(f"{int(p.get_height())}", 
                     (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha='center', va='center', 
                     xytext=(0, 8), 
                     textcoords='offset points', 
                     fontsize=11, fontweight='bold')


# 2. Final clean-up layout adjustments so titles and labels don't overlap
plt.tight_layout()

# 3. Render the unified figure
plt.show()
```
