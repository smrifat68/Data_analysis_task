```python
import pandas as pd
import numpy as np

# Sample Weather Data
data = {
    'Date': pd.to_datetime(['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05']),
    'Temp': [22, np.nan, 25, np.nan, -28],
    'Humidity': [40, -42, np.nan, 45, 46],
    'Event': ['Sunny', 'Sunny', np.nan, 'Rain', 'Cloudy']
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
df = df.replace([-42,-28],np.nan)
```

```python
df
```

```python

df=df.replace({
    'Humidity':np.nan,
   'Event':np.nan,
    'Temp':np.nan
},{
    'Humidity':df.Humidity.median(),
    'Event':'Sunny',
    'Temp':df.Temp.median()
})
```

```python
df
```

```python
df[df.Event=='Sunny'].Temp.max()
```

# Group by using in the Python

```python
g=df.groupby('Event')
g
```

```python
for event, data in g:
    print('evet:',event)
    print('\n')
    print(data)
    print('\n')
```
