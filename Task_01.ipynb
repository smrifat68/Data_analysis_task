# Handle missing values

```python
# Fill Nan value with Median
df['Age'] = df['Age'].fillna(df['Age'].median())
print(df.Age.isnull().sum()) 
# It will show 0 which means there is no Nan value in the Age Column
```

```python
# drop the Cabin column because there isn't enough info to reconstruct it 
df.drop(columns=['Cabin'], inplace=True)
df.head()
```

```python
df['Embarked']=df.Embarked.fillna(df.Embarked.mode()[0]) # Fill with the most common value
print(df.Embarked.isnull().sum()) 
```

# Convert Data Types


```python
# Convert Survived to boolean (True/False) and Embarked to a category
df = df.astype({'Survived':'bool','Embarked':'category'})
```
